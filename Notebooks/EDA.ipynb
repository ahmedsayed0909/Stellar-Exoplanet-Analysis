# EDA

### Importing libraires

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

### Loding Data

```python
data_dir = os.path.abspath(os.path.join(os.getcwd(), '..', 'Data'))
file_path = os.path.join(data_dir, 'cleaned_exoplanet_data.csv')

st_hosts = pd.read_csv(file_path)

st_hosts.head()
```

### The columns ra (Right Ascension) and dec (Declination) map the positions of these systems on the celestial sphere. A scatter plot in equatorial coordinates reveals sky coverage biases (e.g., Kepler/TESS viewing sectors).

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

sns.histplot(st_hosts['st_teff'], bins=50, ax=axes[0], color='royalblue')
axes[0].set_title('Stellar Effective Temperature (st_teff)')
axes[0].set_xlabel('Temperature (K)')

sns.histplot(st_hosts['st_rad'], bins=50, ax=axes[1], color='teal')
axes[1].set_title('Stellar Radius (st_rad)')
axes[1].set_xlabel('Solar Radii')
axes[1].set_xlim(0, 20)  # Zoom in on non-giants for clarity

sns.histplot(st_hosts['sy_dist'], bins=50, ax=axes[2], color='purple')
axes[2].set_title('System Distance (sy_dist)')
axes[2].set_xlabel('Distance (parsecs)')

plt.tight_layout()
plt.show()
```

### Analyze apparent brightness across different filters (sy_vmag, sy_kmag, sy_gaiamag) and their respective uncertainties (err1, err2).

Uncertainty Analysis: Plot error columns (sy_vmagerr1, etc.) to check measurement precision across faint vs. bright targets.

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(
    x='sy_gaiamag',
    y='sy_vmag',
    data=st_hosts,
    alpha=0.3,
    hue='sy_dist',
    palette='viridis',
)
plt.title('Gaia Magnitude vs. V Magnitude')
plt.xlabel('Gaia Magnitude')
plt.ylabel('V Magnitude')
plt.show()
```

### The columns ra (Right Ascension) and dec (Declination) map the positions of these systems on the celestial sphere. A scatter plot in equatorial coordinates reveals sky coverage biases (e.g., Kepler/TESS viewing sectors).

```python
plt.figure(figsize=(12, 6))
sns.scatterplot(
    x='ra', y='dec', data=st_hosts, alpha=0.2, s=5, color='midnightblue'
)
plt.title('Sky Distribution of Targets (RA vs. Dec)')
plt.xlabel('Right Ascension (deg)')
plt.ylabel('Declination (deg)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

### Check how stellar properties correlate with one another (e.g., radius vs. temperature, or distance vs. magnitude).

```python
plt.figure(figsize=(10, 8))
numeric_cols = [
    'st_teff',
    'st_rad',
    'sy_dist',
    'sy_vmag',
    'sy_kmag',
    'sy_gaiamag',
    'sy_pnum',
]
corr = st_hosts[numeric_cols].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Matrix of Stellar & System Parameters')
plt.show()
```
