### Data Cleaning

```python
st_hosts.drop_duplicates(inplace=True)
```

```python
print(st_hosts.isna().sum())
```

```python
import missingno as msno
import matplotlib.pyplot as plt

msno.matrix(st_hosts)
plt.show()
```

```python
# Drop columns with more than 70% missing values
threshold = 0.7 * len(st_hosts)
st_hosts_cleaned = st_hosts.dropna(thresh=threshold, axis=1)

# Drop rows where critical target or ID columns are null
st_hosts_cleaned = st_hosts_cleaned.dropna(subset=['tic_id'])
```

```python
msno.matrix(st_hosts_cleaned)
plt.show()
```

```python
numerical_cols = st_hosts_cleaned.select_dtypes(include=['number']).columns.tolist()
categorical_cols = st_hosts_cleaned.select_dtypes(include=['object', 'category']).columns.tolist()
```

```python
from sklearn.impute import KNNImputer
imputer = KNNImputer(n_neighbors=5)
df_num_imputed = pd.DataFrame(
    imputer.fit_transform(st_hosts_cleaned[numerical_cols]), 
    columns=numerical_cols, 
    index=st_hosts_cleaned.index
)

# 3. Concatenate the imputed numerical features back with the categorical columns
df_final = pd.concat([df_num_imputed, st_hosts_cleaned[categorical_cols]], axis=1)

print(f"Imputation complete! Shape of final dataframe: {df_final.shape}")
```
